## TIme series model for Forecasting

Time series models are statistical tools thath experts use to study and predict data that changes over time. These models help us discover patterns, trends, and relationships in data, which in turn allows us to make informed predictions about what might happen in the future.

Below is a brief overview of some commonly used time series models:

**Moving Average (MA) Model :** This model averages past observations in order to predict future values. It is useful for capturing short-term fluctiations and random variatons in data.
- Assumptions: Observations area linear cobination of past error terms, and there is no autocorrelation between the error terms.
- Parameters: The order of the model (q) determines he number of lagged error terms to include.
- Strenghts: MA models are effective in capturing short-term dependencies and smoothing out random fluctiations in data.

**Autoregressive (AR) model: ** This models predicts future values based on a linear combination of past observations.
- Asssumptions: It assumes that future values depend on past values, capturing long-term trends and dependencies.
- Parameters: The order of the model (p) determines the number of lagged observations to include.
- Strenghts: AR models are useful for capturing long-term trends and dependencies in data.

**Autorregressive Moving Average (ARMA) Model: ** The ARMA model cobines the AR and MA modelss to capture short-term and long-terms patternss in data. It is effective for analyzing stationar time series data.
- Assumptions: Observations are a linear combination of past observations and past error terms, and there is no autocorrelation between the errors terms .
- Parameters: The orders of the AR and MA components (p and q) determine the number of lagged observations and error terms to be included.
- Strengths: modelos combine the streghts of the AR and MA models by capturing both short-term and long-term dependencies in the data

**Autorregressive Integrated Moving Average(ARIMA) Model :** This model extends the AARMA model by incorportaing diffrencing to handle non-stationary data. It is suitable for data with trends or seasonality.

**Seasonal ARIMA(SARIMA) MODEL : ** This model is an extension of the ARIMA model and includes seasonal components. It is useful for analzing and forecasting data with recurring seasonal patterns.
- Assumptions : The data exhibits seasonal patterns as well as trends and dependencies.
- Parameters : The orders of the seasonal AR,I, and MA componentes (P,D and Q) determine the number of lagged seasonal observations, seasonal diffrencing, and laggend seasonal error termms to include.
-Strenghts: SARIMA models are effective for analzing and forecasting time series data with seasoneal patterns



## Difference between a time series and regression problem

Here you might think that since the target varible is numeric, it can be predicted using regression techniques, but a time series problem is different from a regression linear problem in the following ways:
- The main diffrence is that  atime series is time-dependent. Therefore, the basic assumption of alinear regression model that observations are independednt does not apply in the case.
- Along with an incresing or decresing trend, most time series have some form of seasonality trends,i, e., variation specific to a given time period. Therefore, predicting a time series using regression techniques is not a good approach.

Autorregressive models belong to the family of time series model. Theses models capture the relationship between an observation and several lagged observations (previous time steps). Quando aplicamos nosso modelo autorregressivo a dados futuros, o resultado se parece com o mostrado acima. No entanto, podemos aplicar o modelo autorregressivo em tempo real, ou seja, para cada previsão, usamos todos os dados anteriores como dados de treinamento:

```python
window = 24
model = AutoReg(train['Los Angeles'], lags=24)
model_fit = model.fit()
coef = model_fit.params
# walk forward over time steps in test
history = train['Los Angeles'][len(train['Los Angeles'])-window:]
history = [history[i] for i in range(len(history))]
predictions = list()
for t in range(len(valid['Los Angeles'])):
	length = len(history)
	lag = [history[i] for i in range(length-window,length)]
	yhat = coef[0]
	for d in range(window):
		yhat += coef[d+1] * lag[window-d-1]
	obs = valid['Los Angeles'][t]
	predictions.append(yhat)
	history.append(obs)
	print('predicted=%f, expected=%f' % (yhat, obs))
rmse = sqrt(mean_squared_error(valid['Los Angeles'], predictions))
print('Test RMSE: %.3f' % rmse)
```

```python
valid['predictions'] = predictions
plt.plot(valid['Los Angeles'])
plt.plot(valid['predictions'])
plt.show()
```

## Autorregressive Moving Average (ARMA) Model

The ARMA model is a combination of two simpler models: the autorregressive (AR) model and the moving average (MA) model. The ARMA model is used to describe time series daa that are statonary, meaning that their statistical properties do not change over time.

- Autorregressive (AR) model : This model uses the dependence between and observation and a number of lagged observatons (previous time points). It is denoted as AR(p), where p is the number of lagged observations included.
- Moving average (MA) model : This model uses the dependence between an observation and a residual error from a moving average model applied to lagged observations. It is denoted as MA(q), where? q is the number of lagged forecast errors included.

### How to determine the p and q orders in the ARMA model

Determining the appropriate values for p and q is crucial to building an effective ARMA model. This can be done using the following methods:

**Partial Autocorrelation Function (PACF):**

The PACF is used to determine the p-order of the AR model. It measures the correlation between observations at different lags, excluding the influence of intermediate lags.

The p-order is determined by the lag at which the PACF plot is cut.

**Autocorrelation Function (ACF):**

The ACF is used to determine the q-order of the MA model. It measures the correlation between observations at different lags.

The q-order is determined by the lag at which the ACF plot is cut.

To fit a ARMA model, the time series data must be stationary. First, we check for stationary and, if necessary, difference the data to make it stationary. We use the augumented dickey-fuller(adf) test to check for stationary. The ADF test provides a statistic and a p-value. If the p-value is less than 0.05. the series is cconsidered stationary;

```python
from statsmodels.tsa.stattools import adfuller

#Check for stationary
result = adfuller(df_1900['Monthly Anomaly'])
print('ADF Statistic: %f' % result[0])
print('p-value: %f' % result[1])
```

```python
#Let's aply the ARMA model
from statsmodels.tsa.arima.model import ARIMA

model = ARIMA(df_1900['Monthly Anomaly'], order = (2, 0, 2))
model_fit = model.fit()
#Print the sumary model
print(model_fit.summary())
```
